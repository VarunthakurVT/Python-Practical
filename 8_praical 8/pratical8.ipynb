#### Problem 4: Data Loading & Exploration: Create a dataset containing at least 5 columns (e.g., Customer_ID, Name, Age, Gender, Purchase_Amount) and 10 records. Load the dataset into a Pandas DataFrame. Display the first 5 rows using head(). Get summary statistics using describe(). Check for missing values and handle them appropriately. 


```python
customers = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Ethan",
             "Fiona", "George", "Hannah", "Ian", "Julia"],
    "Age": [28, 35, 42, np.nan, 31, 25, 47, 38, 29, 33],
    "Gender": ["Female", "Male", "Male", "Female", "Male",
               "Female", "Male", "Female", None, "Female"],
    "Purchase_Amount": [250.0, 480.5, 320.0, 150.75, np.nan,
                        410.0, 275.25, 530.0, 199.99, 365.5],
})
 
print("Problem 4:")
print("First 5 rows:")
print(customers.head(), end="\n\n")
 
print("Summary statistics:")
print(customers.describe(), end="\n\n")
 
print("Missing values per column:")
print(customers.isnull().sum(), end="\n\n")
 
# Handle missing values (assign back instead of using inplace=True)
customers["Age"] = customers["Age"].fillna(customers["Age"].median())
customers["Purchase_Amount"] = customers["Purchase_Amount"].fillna(
    customers["Purchase_Amount"].mean()
)
customers["Gender"] = customers["Gender"].fillna(
    customers["Gender"].mode()[0]
)
 
print("Missing values after handling:")
print(customers.isnull().sum(), end="\n\n")
```

#### Problem 5: Data Filtering & Selection: Retrieve all customers whose Age is greater than 30. Find the total Purchase Amount for male and female customers separately. Select only the Customer_ID, Name, and Purchase_Amount columns. 

```python

age_above_30 = customers[customers["Age"] > 30]
print("Problem 5: Customers with Age > 30")
print(age_above_30, end="\n\n")
 
#
total_by_gender = customers.groupby("Gender")["Purchase_Amount"].sum()
print("Problem 5: Total Purchase_Amount by Gender")
print(total_by_gender, end="\n\n")
 

selected = customers[["Customer_ID", "Name", "Purchase_Amount"]]
print("Problem 5: Selected columns")
print(selected, end="\n\n")
```

#### Problem 6: Data Transformation & Aggregation: Add a new column "Discounted_Price", where each customer gets a 10% discount on their Purchase_Amount. Group the data by Gender and calculate the average Purchase_Amount. Sort the dataset by Purchase_Amount in descending order.

```python

customers["Discounted_Price"] = customers["Purchase_Amount"] * 0.90
print("Problem 6: With Discounted_Price")
print(customers, end="\n\n")
 
avg_by_gender = customers.groupby("Gender")["Purchase_Amount"].mean()
print("Problem 6: Average Purchase_Amount by Gender")
print(avg_by_gender, end="\n\n")
 
sorted_df = customers.sort_values(by="Purchase_Amount", ascending=False)
print("Problem 6: Sorted by Purchase_Amount (descending)")
print(sorted_df)
```
